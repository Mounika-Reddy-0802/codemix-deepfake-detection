## 3. Rating sheet

Each member scores every clip 1–5 on three axes, then the team agrees the bar.

| Axis | 1 | 5 |
|---|---|---|
| **Speaker similarity** | clearly a different person | indistinguishable from the reference |
| **Code-mix naturalness** | switch points mangled or mispronounced | switches sound native |
| **Artefact-free** | obvious robotic/glitchy artefacts | clean |

(dropped switch, wrong script read aloud, truncation, silence).

**Proposed bar (team confirms):** a clip enters the corpus only if median speaker
similarity ≥ 3 and median artefact-free ≥ 3. Clips failing the bar are logged, not
silently dropped — the failure rate is a number the datasheet reports.
